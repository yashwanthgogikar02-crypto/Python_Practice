```python
class Shape:
    def area(self):
        return "OK"
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
class Rectangle(Shape):
    def __init__(self,length,bredth):
        self.length=length
        self.bredth=bredth
    def area(self):
        return self.length*self.bredth
cir=Circle(2.3)
rec=Rectangle(3,4)
l=[cir,rec]
def find_area(obj):
    print(obj.area())
for i in l:
    find_area(i)
```

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    def fuel_type(self):
        return "Depends on the Vehical"
    @abstractmethod
    def start_engine(self):
        print("Engine _Started")
class Car(Vehicle):
    def fuel_type(self):
        return "This Vehical specifically uses either Disel or petrol"
    def start_engine(self):
        print("Car Engine Started!")
class Bike(Vehicle):
    def fuel_type(self):
        return "This Primarily Runs on Petrol Disel is not Required !"
    def start_engine(self):
        print("Bike Engine Started !")
car=Car()
bike=Bike()
bike.start_engine()
car.start_engine()
a=bike.fuel_type()
print(a)
```
